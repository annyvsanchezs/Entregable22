```python
import numpy as np

# Nombres de medicamentos (array de tipo objeto/string)
nombres = np.array(["Aspirina", "Ibuprofeno", "Diclofenaco", "Acetaminofén", "Naproxeno"], dtype=object)

# identifacion numérica
medicamento_id = np.array([101, 102, 103, 104, 105])

# forma farmaceutica (tableta, cápsula, jarabe)
forma_farmaceutica = np.array(["Tableta", "Tableta", "Cápsula", "Jarabe", "Tableta"], dtype=object)

# Stock (números enteros)
stock = np.array([50, 20, 35, 80, 15])

# Lotes (strings)
lote = np.array(["L001", "L002", "L003", "L004", "L005"], dtype=object)

# Fechas de vencimiento (numpy datetime64)
fecha_venciimiento = np.array(["2026-03-01", "2025-11-15", "2027-01-20", "2024-12-10", "2026-07-30"], dtype="datetime64[D]")

# Mostrar todo
print(nombres)
print(medicamento_id)
print(forma_farmaceutica)
print(stock)
print(lote)
print(fecha_venciimiento)
```

```python
print("Shape stock:", stock.shape)          # (5,)
print("Dtype nombres:", nombres.dtype)      # object
print("Dtype fecha_venciimiento:", fecha_venciimiento.dtype) # datetime
```

```python
print("Primero:", nombres[0], medicamento_id[0], stock[0])
print("Último:", nombres[-1], medicamento_id[-1], stock[-1])
print("Subconjunto:", nombres[1:4])
```

```python
mascara_bajo_stock = stock <= 20
print("Mascara:", mascara_bajo_stock)
print("Medicamentos con stock bajo:", nombres[mascara_bajo_stock])
```

```python
nuevo_stock = stock + 10
print("Stock original:", stock)
print("Stock tras ingreso +10:", nuevo_stock)
# Restar (por venta) 5 unidades de cada producto
stock_venta = stock - 5
print("Stock tras vender 5 unidades:", stock_venta)
stock = stock.copy()
stock = stock - 0
```

```python
print("Total unidades (suma):", np.sum(stock))
print("Promedio de stock:", np.mean(stock))
print("Mediana de stock:", np.median(stock))
print("Stock mínimo:", np.min(stock))
print("Stock máximo:", np.max(stock))
```

```python
indices_orden = np.argsort(fecha_venciimiento)

nombres_orden = nombres[indices_orden]
medicamento_id_orden = medicamento_id[indices_orden]
forma_farmaceutica_orden = forma_farmaceutica[indices_orden]
stock_orden = stock[indices_orden]
lote_orden = lote[indices_orden]
fecha_orden = fecha_venciimiento[indices_orden]

for i in range(len(nombres_orden)):
    print(medicamento_id_orden[i], nombres_orden[i], forma_farmaceutica_orden[i], stock_orden[i], lote_orden[i], fecha_orden[i])
```

```python
dtype = np.dtype([
    ("med_id", "i4"),
    ("nombre", "U20"),
    ("formato", "U10"),
    ("stock", "i4"),
    ("lote", "U10"),
    ("fecha_venciimiento", "datetime64[D]")])

registros = np.empty(len(nombres), dtype=dtype)

registros["med_id"] = medicamento_id
registros["nombre"] = nombres
registros["formato"] = forma_farmaceutica
registros["stock"] = stock
registros["lote"] = lote
registros["fecha_venciimiento"] = fecha_venciimiento

print("Registros (structured array):")
print(registros)
```

```python
nivel_stock = np.empty(len(stock), dtype=object)

nivel_stock[ stock < 20 ] = "Bajo"
nivel_stock[ (stock >= 20) & (stock <= 50) ] = "Medio"
nivel_stock[ stock > 50 ] = "Alto"

print("Niveles:", nivel_stock)
```

```python
formatos_unicos, counts = np.unique(forma_farmaceutica, return_counts=True)
print("Forma farmacéutica:", formatos_unicos)
print("Cuentas por forma farmacéutica:", counts)
```

```python
#Prueba de Fecha de ven
hoy = np.datetime64("today", "D")
vencidos = fecha_venciimiento < hoy
print("Hoy:", hoy)
print("Vencidos:", nombres[vencidos])
```
