```python
import pandas as pd

serie_medicamentos = pd.Series(["Aspirina", "Ibuprofeno", "Diclofenaco", "Acetaminofén", "Naproxeno"], name="Medicamentos")

print(serie_medicamentos)
```

```python
df = pd.DataFrame({
    "medicamento id": [101, 102, 103, 104, 105],
    "nombre": ["Aspirina", "Ibuprofeno", "Diclofenaco", "Acetaminofén","Naproxeno"],
    "formato": ["Tableta", "Tableta", "Cápsula", "Jarabe", "Tableta"],
    "stock": [50, 20, 35, 80, 15],
    "lote": ["L001", "L002", "L003", "L004", "L005"],
    "fecha_vencimiento": ["2026-03-01", "2025-11-15", "2027-01-20", "2024-12-10", "2026-07-30"]})

print(df)
```

```python
def categoria_stock(x):
    if x < 20:
        return "Bajo"
    elif x <= 50:
        return "Medio"
    else:
        return "Alto"

df["nivel_stock"] = df["stock"].apply(categoria_stock)

print(df[["nombre", "stock", "nivel_stock"]])
```
